```python
# x is a local variable, printed inside function as a local variable
def newfunc():
  x = 300
  print(x)

newfunc()
```

```python
# x is a local variable but it is returned to caller
def newfunc():
  x = 300
  return x

print(newfunc())
```

```python
# here x is seen both by the outerfunc and innerfunc
def outerfunc():
  x = 300
  def innerfunc():
    print(x)
  innerfunc()

outerfunc()
```

```python
# x is a global variable so the same x is printed inside and outside

x = 300
def newfunc():
  print(x)

newfunc()
print(x)
```

```python
# here the two 'x' variables are different
x = 300

def newfunc():
  x = 400
  print(x)

newfunc()
print(x)
```

```python
# here x inside the function is a global varible, because of keywork global
def newfunc():
    global x
    x = 660

newfunc()
print(x)
```

```python
# here x variable is nonlocal and belongs to outerfunc, available to innerfunc
# nonlocal: it is neither local nor global.
def outerfunc():
  x = "John"
  def innerfunc():
    nonlocal x
    x = "hello"
  innerfunc()
  return x

print(outerfunc())
```

```python
# use the concept of function and argument passing
# to write a program to display perfect numbers
```

```python
newlist = []
def perfectnumbers():
    for num in range(1,10000):
        sum = 0
        for val in range(1,num):
            if num % val == 0:
                sum = sum + val
        if num == sum:
            newlist.append(num)
    return newlist

x = perfectnumbers()
print(x)
```

```python
def perfectnumbers(b):
    for num in range(1,b):
        sum = 0
        for val in range(1,num):
            if num % val == 0:
                sum = sum + val
        if num == sum:
            print(num,end=" ")

perfectnumbers(10000)
```

# **default function arguments**

```python
#Python program that shows default arguments

def greetings(name="Karthik", msg="Good morning!"):
    print("Hello", name + ', ' + msg)

greetings("Hrishikesh")
greetings("Shalini", "How do you do?")
greetings()
greetings(msg="How do you do")
greetings("","How do you do")
greetings("How do you do")
```

```python
#Python program that shows default arguments

def calculatearea(length=20, breadth=5):
    return length * breadth

print(calculatearea())
print(calculatearea(10))
print(calculatearea(breadth=15))
print(calculatearea(15))
```

```python
#Python program that shows default arguments

def calculatearea(x,length=10,breadth=5):
    return x * length * breadth

# print(calculatearea()) # Error
print(calculatearea(15))
# print(calculatearea(breadth=15)) # Error
print(calculatearea(23,5))
```

```python
num = int(input("Give a number  "))
for trace in range(2,num):
    flag = 0
    for val in range(2,trace):
        if trace % val == 0:
            flag = 1
            break
    if flag == 0:
        print(trace,end=" ")
```
