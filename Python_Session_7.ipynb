```python
# Filter vowel and constant

class LetterFilter:
  def __init__(self, s):
    self.s=s
    self.vowels={"a","e","i","o","u", "A","E","I","O","U"}


  def filter_vowels(self):
  # Remove vowels from string
    return "".join(char for char in self.s if char not in self.vowels)

  def filter_consonants(self):
  # Remove consonants from string
    return "".join(char for char in self.s if char in self.vowels)


s=input()
f= LetterFilter(s);
print(f.filter_vowels())
print(f.filter_consonants())
```

```python
# Give me subset that is equal to k.

def password_strength(password: str, k:int):
  #Edge case: if k is greater than password length, return 0
  if(k>=len(password) or k<=0):
    return 0

  #use a set to collect unique substrings of length k
  seen= set()

  for i in range(len(password)-k+1):
      substring = password[i:i+k]
      seen.add(substring)

  return len(seen)

print(password_strength("suman",2)); #su, um, ma,an
print(password_strength("saurav",2)); #sa, au, ur, ra, av
```

```python
# shopping cart items total price

def shopping_total(items):
  discounted_prices=map(lambda item: item[1]*0.9 if item[1]>50 else item[1], items)



  #calculate total cose
  total= sum(discounted_prices)

  # Round to 1 decimal place
  return round(total, 1)

# Example usage
items=[("Apple", 2.5), ("Banana", 1.8), ("Orange", 3.2), ("Mango", 55.00)];
print(shopping_total(items))

```
