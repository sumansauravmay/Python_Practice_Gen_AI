```python
# List comprehension

print([x*x for x in range(10)]);
```

```python
print([x for x in range(26) if x % 3==0])
```

```python
# Set comprehension
print({x for x in range(10) if x % 2==0})
```

```python
# Dictionary comprehension
print({x:x*x for x in range(10)})
```

```python
# Perfect number in one line
print([x for x in range(1, 10000) if sum(y for y in range(1, x) if x % y==0) ==x])
```

```python
# Map funtion:

def newfunc(x):
  return x*x

expr=list(map(newfunc, [1,2,3,4,5]))
print(expr);
```

```python
myNum=[1,2,3,4,5,6,7,8];
for val in myNum:
  print(val*val, end=" ")
```

```python
# map is function and lambda is iterable
expr=list(map(lambda x:x*x, [1,2,3,4,5]))
print(expr);
```

```python
# Filter
def newFunc(x):
  return x % 5==0

mod=filter(newFunc, range(1,21))
print(list(mod))
```

```python
#Reduce
import functools
sum=functools.reduce(lambda x,y:x+y, range(1,6))
print(sum)
```

```python
import functools
mult=functools.reduce(lambda x,y:x*y, range(1,6))
print(mult)
```

```python
import functools
mult=functools.reduce(lambda x, y: x*y, [3,5,6,10])
print(mult)
```

```python
from itertools import groupby

names=["Alice", "Aaron", "Bob", "Charlie", "Cathy", "Brian"];
#Grouping requires sorting first
names.sort();
print(names);

grouped=groupby(names, key=lambda x:x[0])

for key, group in grouped:
  print(key, list(group));
```

```python
numlist=[2,4,5,6,78,34,22,11];
print(numlist[::]); #it return the list
print(numlist[::-1]); # it reverses the list
print(numlist[::2]); # It jumps two numbers
print(numlist[1]); #it return the 2nd value
print(numlist[-1]); #it return the last value
```
