```python
# from typing import Sequence
# Prine number Sequence 2,3,5,7,11,13,17,19,23
# For loop

for num in range(2, 100):
  flag=0;
  for val in range(2, num):
    if num % val == 0:
      flag=1
      break
  if flag==0:
    print(num, end=" ")
```

```python
# while loop

num=2; val=2;
while num<100:
  flag= 0;
  while(val<num):
    if num%val==0:
      flag=1;
      break;
    val+=1;

  if flag==0:
    print(num, end=" ");
  num=num+1;
  val=2
```
